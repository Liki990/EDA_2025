**map method**
- create a dictionary labels as keys, numbers as values
- in Python index start wiht 0
- case status has two labels we need two values 0 and 1
- Certified becomes 0
- Denied become 1

- we use this method that we are converting categorical to numerical because machien understands only numbers thats the reason.
